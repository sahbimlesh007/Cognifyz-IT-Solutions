### Level 1 : Task 1
1. Top Cuisines Determine the top three most common cuisines in the dataset.
2. Calculate the percentage of restaurants that serve each of the top cuisines.

```python
df['Cuisines'].head(5)
```

- Restaurants can list multiple cuisines (comma-separated), so we split and count each cuisine occurrence individually.

```python
# 1. Top CuisinesDetermine the top three mostcommon cuisines in the dataset.

cusinies_count = df["Cuisines"].str.split(",").explode().str.strip().value_counts()
top_3 = cusinies_count.head(3)
print(f"Top 3 Cuisines are: {top_3}")
```

```python
# 2. Calculate the percentage ofrestaurants that serve each of the top cuisines.

total_res = len(df)

result_set = pd.DataFrame({
    'Top_3_cusinies' : top_3.index,
    'Resturant_count' : top_3.values,
    'Percentage' : (top_3.values/total_res * 100).round(2)
})
print(f'Total number of resturant are: {total_res} \n')
print(f' Percentage of resturants that serve each of the top cuisines are:\n{result_set}')
```

```python
plt.figure(figsize=(8,5))
ax = sns.barplot(x=top_3.index, y=top_3.values, palette="deep")

for container in ax.containers:
    ax.bar_label(container, fmt="%d")

plt.title("Top 3 most common cuisines")
plt.ylabel("Total no. of Resturants")
plt.tight_layout()

plt.savefig('Images/Level_1_assests/Top 3 most common cuisines.png', dpi=500, bbox_inches='tight')

plt.show()
```

**Finding:** 
- North Indian (41.5% of restaurants), Chinese (28.6%), and Fast Food (20.8%) are the three most common cuisines — reflecting the dataset's strong skew toward Indian cities.

### Level 1 : Task 2
1. Identify the city with the highest numberof restaurants in the dataset.
2. Calculate the average rating for restaurants in each city.
3. Determine the city with the highest average rating

```python
# 1. Identify the city with the highest number of restaurants in the dataset.

city_count = df["City"].value_counts()
top_city = city_count.idxmax()
print(f"the city with the highest number of restaurants is {top_city} ({city_count.max()} Resturants)")
```

```python
plt.figure(figsize=(8,4))
city_count.head(5).plot(kind ='bar', color = ['purple', 'pink', 'pink', 'pink', 'pink'])
plt.xticks(rotation = 0)
plt.title('Top 5 cities by no. of Resturants')
plt.ylabel('No of Resturants')
plt.xlabel('')

plt.savefig('Images/Level_1_assests/Top 5 cities by no. of Resturants.png', dpi=500, bbox_inches='tight')

plt.show()
```

```python
# 2. Calculate the average rating for restaurants in each city.

avg_rating = (
    df.groupby("City")["Aggregate rating"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .round(2)
)
print(f"Top 10 city by average rating for all resturants:\n\n{avg_rating}")
```

```python
plt.figure(figsize=(8,4))
ax = sns.barplot(avg_rating, palette='deep')

for container in ax.containers:
    ax.bar_label(container)
    
plt.title('Top 10 city by average rating')
plt.tight_layout()
plt.xticks(rotation = 25)

plt.savefig('Images/Level_1_assests/Top 10 city by average rating.png', dpi=500, bbox_inches='tight')

plt.show()
```

```python
# 3. Determine the city with the highest average rating

reliable_avg_rating = (
    df.groupby("City")["Aggregate rating"]
    .agg(["mean", "count"])
    .query("count >= 20")
    .sort_values("mean", ascending=False)
    .head(10)
)

print(f'Top 10 cities by averge rating (minimum 20 resturants) \n {reliable_avg_rating}')
```

```python
plt.figure(figsize=(10,6))

ax =sns.barplot(x=reliable_avg_rating.index, y=reliable_avg_rating['mean'], palette='Paired')

for container in ax.containers:
    ax.bar_label(container)
    
plt.title('Top 10 Cities by Average Rating (min 20 resturant)')
plt.xlabel('City')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)

plt.savefig('Images/Level_1_assests/Top 10 Cities by Average Rating (min 20 resturant).png', dpi=500, bbox_inches='tight')

plt.show()
```

**Finding:** 

**New Delhi** has by far the most restaurants (5,473  — about 57% of the whole dataset). 

**Inner City** has the highest average rating (4.90), but it only has 2 restaurants, so that's not a reliable signal. 

**Restricting to cities with at least 20 restaurants,** **London** (4.535), **Tampa Bay** (4.410), and **Bangalore** (4.375) top the list — none of them are the dominant New Delhi market, suggesting rating quality doesn't simply track restaurant volume.

### Level 1: Task 3
1. Create a histogram or bar chart to visualize the distribution of price ranges among the restaurants.
2. Calculate the percentage of restaurants in each price range category

```python
# 1. Create a histogram or bar chart to visualize the distribution of price ranges among the restaurants.

total_res = len(df)
price_range_count = df['Price range'].value_counts().sort_index()
print(f'Count by Price Range:\n{price_range_count}')

```

```python
# 2. Calculate the percentage of restaurants in each price range category

price_range_per = df['Price range'].value_counts(normalize=True).sort_index()*100
print(f'Percentage by Price Range:\n{price_range_per.round(2)}')
```

```python
result = pd.DataFrame({
    'Resturant Count' : price_range_count,
    'Percentage(%)' : price_range_per.round(2)
})
print(result)
```

```python
fig, ax = plt.subplots(figsize=(10,7))

ax.bar(price_range_count.index.astype(str), price_range_count.values, color = ['purple', 'pink', 'pink', 'pink'])
ax.set_title('Distribution of Price Range')
ax.set_xlabel('Price Range (low = 1, high = 4)')
ax.set_ylabel('Number of Restaurants')

for i, j in enumerate(price_range_count.values):
    ax.text(i, j+15, f'{j}\n({price_range_per.round(2).values[i]}%)', ha='center', fontsize=8)

plt.tight_layout()

plt.savefig('Images/Level_1_assests/Distribution of Price Range.png', dpi=500, bbox_inches='tight')

plt.show()
    
```

**Finding:**

Nearly half of all restaurants (46.51%) fall in price range 1 (cheapest), and the distribution decreases steadily with price — only 6.14% are in the top price bracket (4). The market is heavily weighted toward budget dining.

### Level 1: Task 4
1. Determine the percentage of restaurants that offer online delivery.
2. Compare the average ratings of restaurants with and without online delivery.

```python
df['Has Online delivery'].unique()
```

```python
# 1. Determine the percentage of restaurantsthat offer online delivery.


online_del_percentage =  df['Has Online delivery'].value_counts(normalize=True).sort_index()*100
print(f'The percentage of resturants that offer online delivery: \n\n{online_del_percentage.round(2)}')
```

```python
ax = sns.barplot(online_del_percentage.round(2), palette='dark')

for container in ax.containers:
    ax.bar_label(container)
plt.title('Percentage of resturant offering online delivery')
plt.xlabel('')
plt.ylabel('in (%)')
plt.tight_layout()

plt.savefig('Images/Level_1_assests/Percentage of resturant offering online delivery.png', dpi=500, bbox_inches='tight')

plt.show()
```

```python
# 2. Compare the average ratings of restaurants with and without online delivery.


avg_delivery_Isonline = df.groupby('Has Online delivery')['Aggregate rating'].mean().round(2)
print(f'Average rating of resturant with and without online delivery:\n\n {avg_delivery_Isonline} ')
```

```python
ax = sns.barplot(avg_delivery_Isonline, palette='dark')

for container in ax.containers:
    ax.bar_label(container)
plt.title('AVG Rating: Online vs Not')
plt.xlabel('')
plt.ylabel('Avg. Rating')
plt.tight_layout()

plt.savefig('Images/Level_1_assests/AVG Rating_Online vs Not.png', dpi=500, bbox_inches='tight')

plt.show()
```

**Finding:** 

Only 25.69% of restaurants provide online delivery. However, restaurants that offer online delivery have a higher average rating (3.25) compared to those that do not (2.46), a difference of about 0.79 stars.

This suggests that restaurants with online delivery tend to be better rated, possibly because larger or better-managed restaurants are more likely to offer delivery services.
