### Level 1 : Task 1
1. Top Cuisines Determine the top three mostcommon cuisines in the dataset.
2. Calculate the percentage ofrestaurants that serve each of the top cuisines.

```python
df['Cuisines'].head(5)
```

```python
cusinies_count = df['Cuisines'].str.split(',').explode().str.strip().value_counts()
```

```python
# 1. Top CuisinesDetermine the top three mostcommon cuisines in the dataset.


top_3 = cusinies_count.head(3)
top_3
```

```python
total_res = len(df)
```

```python
# 2. Calculate the percentage ofrestaurants that serve each of the top cuisines.


result_set = pd.DataFrame({
    'Top_3_cusinies' : top_3.index,
    'Resturant_count' : top_3.values,
    'Percentage' : (top_3.values/total_res * 100).round(2)
})
```

```python
result_set
```

### Level 1 : Task 2
1. Identify the city with the highest numberof restaurants in the dataset.
2. Calculate the average rating for restaurants in each city.
3. Determine the city with the highest average rating

```python
df.columns
```

```python
# 1. Identify the city with the highest numberof restaurants in the dataset.


highest_res_city = df.groupby('City')['Restaurant Name'].count().sort_values(ascending=False).head(1)
print(highest_res_city)
```

```python
# 2. Calculate the average rating for restaurants in each city.


avg_rating = df.groupby('City')['Aggregate rating'].mean().sort_values(ascending=False)
print(avg_rating.round(2))
```

```python
# 3. Determine the city with the highest average rating


highest_avg_city = avg_rating.head(1)
print(highest_avg_city)
```

### Level 1: Task 3
1. Create a histogram or bar chart to visualize the distribution of price ranges among the restaurants.

```python
# 1. Create a histogram or bar chart to visualize the distribution of price ranges among the restaurants.


price_range_count = df['Price range'].value_counts().sort_values(ascending=False)
```

```python
sns.barplot(price_range_count)
```

```python
price_count = df['Price range'].value_counts().sort_index()
price_count
```

```python
# 2. Calculate the percentage of restaurants in each price range category

price_percentage = df['Price range'].value_counts(normalize=True).sort_index()*100
price_percentage
```

```python
result = pd.DataFrame({
    'Resturant Count' : price_count,
    'Price Percentage(%)' : price_percentage
})
```
