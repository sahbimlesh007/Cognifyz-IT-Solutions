### Level 3: Task 1
1. Analyze the text reviews to identify the mostcommon positive and negative keywords.
2. Calculate the average length of reviews andexplore if there is a relationship between review length and rating.


```python
df.columns
```

```python
df['Rating text'].head(5)
```

```python
df['Rating text'].unique()
```

*There is no specific column to analyse the review. But here, I am using Rating Text as the review*

```python
df['Rating text'].value_counts()
```

```python
# 1. Analyze the text reviews to identify the mostcommon positive and negative keywords.

positive = df[df['Aggregate rating']>=3]['Rating text'].value_counts()
negative = df[df['Aggregate rating']<3]['Rating text'].value_counts()
print(f'Positive Review: {positive}')
print(f'Negative Review: {negative}')
```

```python
df.columns
```

```python
# 2. Calculate the average length of reviews and explore if there is a relationship between review length and rating.

df['review_length'] = df['Rating text'].astype(str).str.len()

df['review_length'].head()
```

```python
df.groupby('Rating text')['review_length'].mean()
```

```python
avg_len = round(df['review_length'].mean(),2)
print(f'Average lenght of the review is {avg_len}')
```

```python
corr_relation = df['review_length'].corr(df['Aggregate rating'])
corr_relation
```

```python
if corr_relation <= 1 and corr_relation > 0:
    print('there is a relationship between review length and rating. Longer  reviews associated with higher ratings')
    
elif corr_relation < 0 and corr_relation >= -1 :
    print('there is a relationship between review length and rating. longer reviews associated with lower ratings')
    
elif corr_relation == 0:
    print('there is a relationship between review length and rating.')
```

### Level 3: Task 2
1. Identify the restaurants with the highest and lowest number of votes.
2. Analyze if there is a correlation between the number of votes and the rating of a restaurant

```python
df['Aggregate rating'].unique()
```

```python
df.columns
```

```python
# 1. Identify the restaurants with the  lowest number of votes.

lowest_votes_resturant = df.loc[df['Votes'].idxmin(), ['Restaurant Name', 'Votes']]
lowest_votes_resturant
```

```python
# 1. Identify the restaurants with the highest  number of votes.

highest_votes_resturant = df.loc[df['Votes'].idxmax(), ['Restaurant Name', 'Votes']]
highest_votes_resturant
```

```python
# 2. Analyze if there is a correlation between the number of votes and the rating of a restaurant

df['Votes'] = df['Votes'].astype(float)
df['Aggregate rating'] = df['Aggregate rating'].astype(float)
```

```python
corr_relation = df['Votes'].corr(df['Aggregate rating'])
corr_relation
```

```python
sns.scatterplot(data = df, x = 'Votes', y = 'Aggregate rating')
sns.regplot(data=df, x='Votes', y='Aggregate rating',scatter_kws={'alpha':0.5})
```

### Level 3: Task 3
1. Analyze if there is a relationship between theprice range and the availability of online delivery and table booking.
2. Determine if higher-priced restaurants aremore likely to offer these services

```python
df.columns
```

```python
df['Has Online delivery'].unique()
```

```python
# 1. Analyze if there is a relationship between the price range and the availability of online delivery and table booking.


rel_price_onDelivery = pd.crosstab(df['Price range'], df['Has Online delivery'], normalize='index') * 100
rel_price_onDelivery
```

```python
df['Has Table booking'].unique()
```

```python
rel_price_tableBooking = pd.crosstab(df['Price range'],df['Has Table booking'],normalize='index') * 100
rel_price_tableBooking
```

```python
sns.countplot(data=df, x='Price range', hue='Has Online delivery')
```

```python
sns.countplot(data=df, x='Price range', hue='Has Table booking')
```

```python
# 2. Determine if higher-priced restaurants aremore likely to offer these services


service_analysis = df.groupby("Price range").agg(
    Online_del=("Has Online delivery", lambda x: (x == "Yes").mean() * 100),
    Table_booking=("Has Table booking", lambda x: (x == "Yes").mean() * 100),
)

service_analysis
```
