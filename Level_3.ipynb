### Level 3: Task 1
1. Analyze the text reviews to identify the most common positive and negative keywords.
2. Calculate the average length of reviews and explore if there is a relationship between review length and rating.


### *The DataSet has no specific column to analyse the review. But here, I am using Rating Text as the review*

```python
df['Rating text'].unique()
```

```python
df['Rating text'].value_counts()
```

```python
# 1. Analyze the text reviews to identify the most common positive and negative keywords.

positive = df[df['Aggregate rating']>=3]['Rating text'].value_counts()
negative = df[df['Aggregate rating']<3]['Rating text'].value_counts()
print(f'Positive Review:\n{positive}')
print()
print(f'Negative Review:\n{negative}')
```

```python
# 2. Calculate the average length of reviews and explore if there is a relationship between review length and rating.

df['review_length'] = df['Rating text'].astype(str).str.len()
df.groupby('Rating text')['review_length'].mean()
avg_len = round(df['review_length'].mean(),2)
print(f'Average lenght of the review is {avg_len}')

corr_relation = df['review_length'].corr(df['Aggregate rating'])
print(f'Correlation b/w review length and rating is {corr_relation}')
```

```python
review_df = pd.DataFrame({
    'Positive': positive,
    'Negative': negative
}).fillna(0)

ax = review_df.plot(kind='bar', figsize=(9,6), width = 0.9)

for container in ax.containers:
    ax.bar_label(container)

plt.title('Review Analysis (+ve vs -ve)')
plt.xlabel('')
plt.ylabel('No. of Reviews')
plt.xticks(rotation = 0)
plt.tight_layout()

plt.savefig('Images/Level_3_assests/Review Analysis.png', dpi = 500, bbox_inches = 'tight')

plt.show()
```

**Finding:**
- Most of the resturants gained positive review concentrated more around average(2487), good( 2096) followed by very good and excellent. However, the large number of Not Rated (2,148) restaurants suggests that many establishments have limited customer feedback and engagement.
- The correlation of -0.47 shows that there is a negative relationship between review length and rating means longer reviews associated with lower ratings

### Level 3: Task 2
1. Identify the restaurants with the highest and lowest number of votes.
2. Analyze if there is a correlation between the number of votes and the rating of a restaurant

```python
df['Aggregate rating'].unique()
```

```python
# 1. Identify the restaurants with the highest  number of votes.

highest_votes_resturant = df.loc[df['Votes'].idxmax(), ['Restaurant Name', 'Votes', 'Aggregate rating']]

lowest_votes_resturant = df.loc[df['Votes'].idxmin(), ['Restaurant Name', 'Votes','Aggregate rating']]

print(f'The restaurant with highest number of votes is:\n{highest_votes_resturant} ')
print()
print(f'The restaurant with highest number of votes is:\n{lowest_votes_resturant} ')
```

```python
# 2. Analyze if there is a correlation between the number of votes and the rating of a restaurant

corr_relation = df['Votes'].corr(df['Aggregate rating'])

print(f"Correlation between votes and rating: {corr_relation:.3f}")
```

```python

plt.figure(figsize=(6,4))
sns.scatterplot(data = df, x = 'Votes', y = 'Aggregate rating', alpha = 0.3)

plt.title(f'Votes vs Rating (correlation = {corr_relation:.2f})')
plt.xscale('symlog')
plt.xlim(left=-1)
plt.tight_layout()

plt.savefig('Images/Level_3_assests/Corr of voting vs rating.png', dpi = 500, bbox_inches = 'tight')

plt.show()
```

**Finding:** 

**Toit** (a Bangalore brewery/restaurant) has the most votes (10,934) with a 4.8 rating. There's a weak-to-moderate positive correlation (0.313) between votes and rating — more-voted restaurants tend to be rated a bit higher, but plenty of highly rated restaurants have few votes, and the relationship is far from strict.

### Level 3: Task 3
1. Analyze if there is a relationship between the price range and the availability of online delivery and table booking.
2. Determine if higher-priced restaurants are more likely to offer these services

```python
df['Has Online delivery'].unique()
```

```python
df['Has Table booking'].unique()
```

```python
# 1. Analyze if there is a relationship between the price range and the availability of online delivery and table booking.


delivery_by_price = pd.crosstab(df['Price range'], df['Has Online delivery'], normalize='index') * 100

booking_by_price = pd.crosstab(df['Price range'],df['Has Table booking'],normalize='index') * 100

print("% offering online delivery, by price range:\n", delivery_by_price.round(1))
print()
print("% offering table booking, by price range:\n", booking_by_price.round(1))
```

```python
# 2. Determine if higher-priced restaurants are more likely to offer these services


service_analysis = df.groupby("Price range").agg(
    Online_del=("Has Online delivery", lambda x: (x == "Yes").mean() * 100),
    Table_booking=("Has Table booking", lambda x: (x == "Yes").mean() * 100),
).round(1)

service_analysis
```

```python
plt.figure(figsize=(8,5))

ax = service_analysis.plot(kind='bar', figsize=(8,5), width = 0.7)

for container in ax.containers:
    ax.bar_label(container)

plt.title('Online Delivery and Table Booking by Price Range')
plt.xlabel('Price Range')
plt.ylabel('Percentage (%)')
plt.legend(['Online Delivery', 'Table Booking'], title='Services')
plt.tight_layout()
plt.xticks(rotation = 0)

plt.savefig('Images/Level_3_assests/Online Delivery and Table Booking by Price Range.png', dpi = 500, bbox_inches = 'tight')
plt.show()
```

**Finding:** 

The two services follow **opposite** trends across price range:

- **Table booking** rises sharply with price — from 0% at price range 1 to ~46–47% at price ranges 3 and 4. Higher-priced restaurants are far more likely to offer table booking (makes sense — reservations suit sit-down dining).
- **Online delivery** peaks in the *middle* (41.3% at price range 2) and actually **drops** at the highest price range (9%). Budget and mid-range restaurants are much more delivery-oriented than fine dining.

So higher-priced restaurants are more likely to offer table booking, but *less* likely to offer online delivery — the opposite of a simple "higher price = more services" pattern.
